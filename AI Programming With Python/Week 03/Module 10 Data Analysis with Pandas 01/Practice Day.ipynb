7. Create a DataFrame of 8 students with columns "Name" and "Score". Filter and print students with Score > 60 and those with Score < 40.

```python
data = {
    "Name": ["Akib", "Rahim", "Karim", "Sakib", "Naim", "Jahid", "Rafi", "Tanvir"],
    "Score": [85, 78, 92, 88, 76, 81, 90, 30]
}

df = pd.DataFrame(data)
df
```

```python
df_greater_60 = df.loc[(df['Score'] > 60)]
df_greater_60
```

```python
df_less_than_40 = df.loc[df['Score'] < 40]
df_less_than_40
```

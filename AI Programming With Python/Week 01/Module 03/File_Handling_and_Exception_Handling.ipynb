Practice 1 :
Get the text from sample.txt , and count the following

Number of lines

Number of words

Number of characters

and save it to the counter_of_string.txt file

```python
from functools import reduce
with open("./sample_data/sample.txt", "r") as file :
  string_list = file.readlines()
  # print(strings)

  # total line
  total_lines = len(string_list)

  print(total_lines)

  # total word
  number_of_words = list(map(lambda x : len(x.split()), string_list))

  print(number_of_words)

  total_number_of_words = reduce(lambda x, y : x + y, number_of_words)

  # total words
  print(total_number_of_words)

  # print(string_list)


  # cleaning process

  # 1. new line delete
  string_list = list(map(str.strip, string_list))

  print(string_list)

  # 2. Space delete
  string_list = list(map(lambda x : x.replace(" ", ""), string_list))

  # print(string_list)

  # 3. How many characters
  total_number_of_characters = list(map(lambda x : len(x), string_list))

  print(total_number_of_characters)

  total_number_of_characters = reduce(lambda x, y : x + y, total_number_of_characters)

  # total character
  print(total_number_of_characters)

with open("./sample_data/counter_of_string.txt", "w") as file :
  file.write(f"Total lines : {total_lines}\nTotal Number of words : {total_number_of_words}\nTotal Number of chartacters : {total_number_of_characters}")

```

1. Create a text file named emails.txt containing multiple email addresses, some in uppercase and some repeated.
Write a Python program to clean the data by converting all emails to lowercase, removing duplicates, sorting them alphabetically, and saving the cleaned list to a new file named unique_emails.txt.

```python
with open("emails.txt", "r") as file :
  emails = file.readlines()

# print(emails)
emails = list(map(lambda x : x.strip().lower(), emails))
# emails

lst = []

for i in emails :
  if i not in lst :
    lst.append(i)

# lst

with open("unique_emails.txt", 'a') as file :
  for i in lst :
    file.write(i+"\n")

```

2.Create a text file named names.txt containing a list of names.
Some names may have extra spaces, inconsistent capitalization (like “aLiCe” or “ bob ”), or appear more than once.
Write a Python program to clean the data by:
1.  Removing extra spaces.
2.  Converting all names so that only the first letter is capitalized.
3.  Removing duplicate names.
4. Sorting the names alphabetically.
Finally, save the cleaned list into a new file named clean_names.txt.

```python
with open("names.txt", "r") as file :
  names = file.readlines()

names = list(map(lambda x : x.strip().capitalize(), names))
# print(names)

lst = []
for i in names :
  if i not in lst :
    lst.append(i)

# lst
lst.sort()

with open("clean_names.txt", "a") as file :
  for i in lst :
    file.writelines(i + "\n")
```

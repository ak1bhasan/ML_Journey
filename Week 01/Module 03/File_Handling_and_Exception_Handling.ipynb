Practice 1 :
Get the text from sample.txt , and count the following

Number of lines

Number of words

Number of characters

and save it to the counter_of_string.txt file

```python
from functools import reduce
with open("./sample_data/sample.txt", "r") as file :
  string_list = file.readlines()
  # print(strings)

  # total line
  total_lines = len(string_list)

  print(total_lines)

  # total word
  number_of_words = list(map(lambda x : len(x.split()), string_list))

  print(number_of_words)

  total_number_of_words = reduce(lambda x, y : x + y, number_of_words)

  # total words
  print(total_number_of_words)

  # print(string_list)


  # cleaning process

  # 1. new line delete
  string_list = list(map(str.strip, string_list))

  print(string_list)

  # 2. Space delete
  string_list = list(map(lambda x : x.replace(" ", ""), string_list))

  # print(string_list)

  # 3. How many characters
  total_number_of_characters = list(map(lambda x : len(x), string_list))

  print(total_number_of_characters)

  total_number_of_characters = reduce(lambda x, y : x + y, total_number_of_characters)

  # total character
  print(total_number_of_characters)

with open("./sample_data/counter_of_string.txt", "w") as file :
  file.write(f"Total lines : {total_lines}\nTotal Number of words : {total_number_of_words}\nTotal Number of chartacters : {total_number_of_characters}")

```

### Problem 1  Create and Write to a File

```python
def create_file(filename, content) :
  with open(filename, "w") as file :
    file.write(content)

create_file("notes.txt", "Hello, this is my first file.")
```

### Problem 2  Read Lines with Line Numbers

```python
def read_lines(filename) :
  with open(filename, "r") as file :
    i = 1
    for line in file :
      print(f"{i} : {line}", end = "")
      i += 1
read_lines("data.txt")
```

### Problem 3  Append Log Entries to a File

```python
def add_log(message) :
  with open("log.txt", "a") as file :
    file.write(message)

add_log("User logged in\n")
add_log("File uploaded\n")

```

### Problem 4  Read from a Specific File Pointer Position

```python
def read_from_position(filename, position) :
  with open(filename, "r") as file :
    file.seek(position)
    print(file.read(position))

read_from_position("hello.txt", 7)
```

### Problem 5  ⭐ Conceptual Session — Safe File Read with Full Exception Handling

```python
def safe_read(filename) :
  try :
    with open(filename, "r") as file :
      print(file.read())
  except FileNotFoundError as e :
    print("Error: File not found.")
  finally :
    print("Operation complete.")

safe_read("missing.txt")
safe_read("hello.txt")
```

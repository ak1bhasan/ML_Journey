### Problem 2

```python
def batch_generator( d, b ) :
    for i in range( 0, len(d), b ) :
        print(d[ i : i + b ])

data = []
for i in input().split() :
    data.append(int(i))
    
# print(data)

batch_size = int(input())
# batch_size

batch_generator( data, batch_size ) 
```

### Problem 4

```python
def power( b, e = None ) :
    if e != None :
        print( b ** e )
    else :
        print( b ** 2 )

inp = []
for i in input().split() :
    inp.append(int(i))

# inp

if len(inp) == 1 :
    power( inp[0], None )
else :
    power( inp[0], inp[1] )
```

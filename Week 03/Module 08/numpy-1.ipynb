## Ndarray

```python
import numpy as np

# 1d array

arr1 = np.array([1, 2, 3, 4, 5])

print(type(arr1))
print(arr1.ndim)

# 2d array

arr2 = np.array([ [1, 2, 3], # row 1
                  [4, 5, 6] # row 2
                ]) 

print(arr2.ndim)


# 3d array

arr3 = np.array([ 
    # 1st floor
    [ [1, 2, 3], # row 1
    [4, 5, 6] ], # row 2
    
    # 2nd floor
    [ [1, 2, 3], # row 1
    [4, 5, 6] ], # row 2

    # 3rd floor
    [ [1, 2, 3], # row 1
    [4, 5, 6] ] # row 2
    
    ]) 

print(arr3.ndim)

```

## Array Creation with range functions

```python
# np.arange(first, end, step) -> similar to range function

arr = np.arange(1, 10, 1)
mat = arr.reshape(3, 3)

print(arr)
print(mat)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.linspace(start, end(included), koita)

arr = np.linspace(1, 10, 20)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.logspace(start, end(included), koita)

arr = np.logspace(0, 4, 6, base = 2)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

1. You have a list of temperatures recorded in Celsius. Your task is to:
Create a NumPy array from the given list.
Convert all temperatures from Celsius to Fahrenheit. The formula is F = C* 1.8 + 32.
Print the new Fahrenheit array.

```python
temp_cel = [-50, -20, 0, 5, 10, 15, 20, 25, 30, 35, 40, 50]

temp_cel = np.array(temp_cel)

print(type(temp_cel))

temp_Fahrenheit = (temp_cel * 1.8 + 32)

print(temp_Fahrenheit)
print(type(temp_Fahrenheit))
print(temp_Fahrenheit.dtype)
```


2.  You have a Python list containing the lap times (in seconds) for an athlete.

Task:

Convert the list into a NumPy array.
Print the array's shape, size, and data type.
The athlete wants to see their times in minutes. Create a new array by dividing all the lap times by 60.
Print the new array of lap times in minute


```python
lap_times = [72, 70, 69, 71, 68, 70, 69]

lap_times = np.array(lap_times)

print(lap_times.shape)
print(lap_times.size)
print(lap_times.dtype)

new_arr = lap_times / 60
print(new_arr)
```

## Creating ndarray from Scratch

```python
arr1 = np.array([1, 2, 3, 4, 5])

print(type(arr1))
print(arr1.ndim)

# 2d array

arr2 = np.array([ [1, 2, 3], # row 1
                  [4, 5, 6] # row 2
                ]) 

print(arr2.ndim)


# 3d array

arr3 = np.array([ 
    # 1st floor
    [ [1, 2, 3], # row 1
    [4, 5, 6] ], # row 2
    
    # 2nd floor
    [ [1, 2, 3], # row 1
    [4, 5, 6] ], # row 2

    # 3rd floor
    [ [1, 2, 3], # row 1
    [4, 5, 6] ] # row 2
    
    ]) 

print(arr3.ndim)
```

```python
# np.zeros

arr = np.zeros((2,3), dtype = np.int8)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
arr = np.zeros_like(arr3) ## Same as arr3

print(arr)
print(arr.shape)
```

```python
# np.ones

arr = np.ones((3,4,3), dtype = np.int8)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
arr = np.ones_like(arr3) ## Same as arr3

print(arr)
print(arr.shape)
```

```python
# np.empty

arr = np.empty((4,3), dtype = np.int8)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
arr = np.empty_like(arr3) ## Same as arr3

print(arr)
print(arr.shape)
```

```python
# np.full

arr = np.full((4,3, 3), np.inf)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
arr = np.full_like(arr3, 10.5) ## Same as arr3

print(arr)
print(arr.shape)
print(arr.dtype)
```

```python
arr = np.full_like(arr3, np.inf, dtype = np.float64) ## Same as arr3

print(arr)
print(arr.shape)
```

## Creating matrix for Linear algebra

```python
diagonal_matrix = np.diag([1, 2, 3, 4])

print(diagonal_matrix)
print(diagonal_matrix.shape)
```

```python
# Identity matrix

identity_mat = np.eye(3)
print(identity_mat)
print(identity_mat.shape)
```

```python
# np.eye(row, column)

mat = np.eye(3, 4)
print(mat)
```

```python
# np.eye(row, column, k)

mat = np.eye(3, 4, -1)
print(mat)
```

```python
mat = np.eye(3, 4, 1)
print(mat)
```

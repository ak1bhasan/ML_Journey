# Solution :

## 1. Difference between AI and ML :

    AI : A broad field that aims to make machines perform tasks requiring human intelligence.

    ML : A subset of AI that enables machines to learn from data and improve automatically.

    AI is a broader concept, while ML is a technique used within AI.


## 2. Difference between ANI and AGI :  

    ANI(Artificial Narrow Intelligence) :

        *   Designed for a specific task.
        *   Example: Voice assistants, recommendation systems.
    
    AGI(Artificial General Intelligence) :

        *   Can perform any intellectual task that a human can do.
        *   Does not yet exist.
    
    ANI is specialized for one task, while AGI would have human-like general intelligence.

## 3. Deep Learning uses neural networks to learn patterns from data.

    Since it is a learning technique within Machine Learning, it is considered a subset of ML.

    Hierarchy: AI → ML → Deep Learning

## 4. Alan Turing is considered one of the founders of AI.

        *   Inspired early AI research.

## Types of Machine Learning
1. Supervised Learning কী?
2. Regression এবং Classification এর মধ্যে পার্থক্য লিখো।
3. House Price Prediction কোন ধরনের problem? কেন?
4. Spam Email Detection কোন ধরনের ML problem?
5. Unsupervised Learning এ labels কেন থাকে না?


# Solution

## 1. What is Supervised Learning?

Supervised Learning is a type of Machine Learning where the model learns from labeled data.

The dataset contains both input features and correct output labels.

**Example:** House Price Prediction, Spam Email Detection.

## 2. Difference between Regression and Classification

### Regression

- Predicts a continuous numerical value.
- Example: House price, temperature prediction.

### Classification

- Predicts a category or class.
- Example: Spam/Not Spam, Disease/No Disease.

**Answer:** Regression predicts numbers, while Classification predicts categories.

## 3. What type of problem is House Price Prediction? Why?

House Price Prediction is a **Regression** problem.

Because the output is a continuous numerical value (house price).

## 4. What type of ML problem is Spam Email Detection?

Spam Email Detection is a **Classification** problem.

Because the model classifies emails into categories such as Spam or Not Spam.

## 5. Why are there no labels in Unsupervised Learning?

In Unsupervised Learning, the dataset contains only input data and no target labels.

The model tries to find hidden patterns, relationships, or groups within the data by itself.

**Answer:** Labels are not available because the goal is to discover patterns from unlabeled data.


## ML Pipeline
1. Feature Engineering কী?
2. Train-Test Split কেন প্রয়োজন?
3. Data Preprocessing এর ৩টি কাজ লিখো।
4. Hyperparameter এবং Parameter এর মধ্যে পার্থক্য কী?
5. Model Drift বলতে কী বুঝায়?


# Solution :

## 1. What is Feature Engineering?

Feature Engineering is the process of creating, selecting, and transforming features from raw data to improve model performance.

It helps the model learn better patterns from data.

## 2. Why is Train-Test Split needed?

Train-Test Split is needed to evaluate model performance on unseen data.

- Train set: Used to train the model
- Test set: Used to check model accuracy

**Answer:** It helps prevent overfitting and measures real-world performance.

## 3. Write 3 tasks of Data Preprocessing

1. Handling missing values (imputation or removal)  
2. Encoding categorical data (Label Encoding / One-Hot Encoding)  
3. Feature scaling (Normalization / Standardization)

## 4. Difference between Hyperparameter and Parameter

### Parameter
- Learned by the model during training
- Example: weights in linear regression

### Hyperparameter
- Set manually before training
- Example: learning rate, number of trees

**Answer:** Parameters are learned from data, while hyperparameters are set by the user.

## 5. What is Model Drift?

Model Drift happens when a machine learning model’s performance decreases over time because the data distribution changes.

**Example:** A spam detection model becomes less accurate when new types of spam emails appear.

**Answer:** It is the degradation of model performance due to changing real-world data.

## Simple Linear Regression
1. Simple Linear Regression কী predict করে?
2. Independent Variable এবং Dependent Variable এর পার্থক্য লিখো।
3. y = mx + b equation এ m এবং b কী বোঝায়?
4. Best Fit Line বলতে কী বুঝায়?
5. Linear Regression এর একটি real-life use case লিখো।


# Solution :

## 1. What does Simple Linear Regression predict?

Simple Linear Regression predicts a **continuous numerical value** based on one input variable.

**Example:** Predicting house price from size.


## 2. Difference between Independent and Dependent Variable

### Independent Variable (X)
- Input variable
- Used to make prediction
- Example: House size

### Dependent Variable (Y)
- Output variable
- Value we want to predict
- Example: House price

**Answer:** Independent variable is input, dependent variable is output.

## 3. In y = mx + b, what are m and b?

- **m (slope):** Shows how much y changes when x changes
- **b (intercept):** Value of y when x = 0

## 4. What is Best Fit Line?

Best Fit Line is the straight line that represents the relationship between input and output data with minimum error.

It minimizes the distance between actual and predicted values.

## 5. Real-life use case of Linear Regression

Predicting **house prices** based on area (square feet).

Other examples:
- Salary prediction based on experience
- Temperature prediction over time

## Cost Function
1. Cost Function কেন দরকার?
2. Prediction Error কী?
3. Model error কমানো কেন গুরুত্বপূর্ণ?
4. Large error হলে Cost Function এর value কেমন হবে?
5. Cost কমানোর process এর নাম কী?


# Solution :

## 1. Why is Cost Function needed?

Cost Function is needed to measure how wrong the model’s predictions are compared to actual values.

It helps in evaluating model performance.

## 2. What is Prediction Error?

Prediction Error is the difference between the actual value and the predicted value.

**Error = Actual value − Predicted value**

## 3. Why is it important to reduce model error?

Reducing model error improves accuracy and makes the model more reliable for real-world predictions.

## 4. What happens to Cost Function when error is large?

If the error is large, the Cost Function value will also be large.

This shows the model is performing poorly.

## 5. What is the process of minimizing cost called?

The process of minimizing the Cost Function is called **Gradient Descent**.

# Task 6:
• Write the code to calculate the percentage distribution of the island column

```python
( df['island'].value_counts() / len(df) ) * 100 
```

```python
distribution = df['island'].value_counts()

distribution_label = df['island'].unique()

# print(distribution_label)

plt.pie(distribution, labels = distribution_label, autopct = '%1.2f%%')
plt.show()
```

# Task 7:
Let's visualize the gender distribution among the surveyed penguins.

• Extract the value counts for the sex column.

• Create a pie chart using Matplotlib showing the male vs. female distribution.

• Ensure the pie chart includes percentage labels with one decimal point (example 50.5%).

Add a title to your chart.

```python
df.sample(5)
```

```python
df['sex'].value_counts()
```

```python
df['sex'].isnull().sum()

df.dropna(subset = ['sex'], inplace = True)
```

```python
sex_counts = ( df['sex'].value_counts() / len(df) ) * 100

# print(value_counts)

sex_label = df['sex'].unique()
# print(sex_label)

plt.pie(sex_counts, labels = sex_label, autopct = '%1.1f%%')
plt.title('Gender Distribution of Penguins')
plt.show()
```

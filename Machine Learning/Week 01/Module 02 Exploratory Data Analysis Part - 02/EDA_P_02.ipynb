```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('titanic_data_updated.csv')
df.head()
```

# Multivariate Analysis

## 2.2 & 2.3 Multivariate Analysis- 1 ( Countplot with Hue )

```python
sns.countplot(x = df['Sex'], hue = df['Survived'])
plt.show()
```

```python
gender_split = df.groupby('Sex')['Survived'].value_counts(normalize = True) # normalize=True returns proportions (probabilities instead of counts)

print(gender_split)
```

```python
sns.countplot(x = df['Pclass'], hue = df['Survived'])
plt.show()
```

```python
pclass_split = df.groupby('Pclass')['Survived'].value_counts(normalize = True)

print(pclass_split)
```

# 2.4 Multi Variate Analysis -3 ( Barplot for Cat-Num)

```python
sns.barplot(x = df['Pclass'], y = df['Fare'])
plt.show()
```

```python
sns.barplot(x=df['Survived'],y=df['Age'])
plt.show()
```

```python
sns.barplot(x = df['Survived'], y = df['Age'], hue = df['Sex'])
plt.show()
```

# 2.6 Multivariate Analysis -4 ( KDE plot for categorical discrepancy )

```python
sns.kdeplot(data = df, x = df['Age'], hue = 'Survived')

plt.show()
```

```python
sns.kdeplot(data = df, x = 'Age', hue = 'Pclass')

plt.show()
```

```python
sns.kdeplot(data = df, x = 'Fare', hue = 'Survived')

plt.show()
```

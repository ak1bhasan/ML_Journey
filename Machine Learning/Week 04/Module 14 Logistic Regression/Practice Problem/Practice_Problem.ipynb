# Problem 03

```python
def sigmoid( z ) :
  return ( 1 / ( 1 + np.exp( -z ) ) )


W = [2.0, 2.0]
b = -1.0

X = np.array([[1.5, 0.5]])
y = 1

z = np.dot( W, X[0] ) + b

prob = sigmoid( z )

error = prob - y
dj_dw = error * X[0]

print(f"Compute z (np.dot(W, X[0]) + b) : {round(z, 2)}")
print(f"Predicted probability (prob)    : {round(prob, 2)}")
print(f"Error vector (prob - y[0])      : {round(error, 2)}")
print(f"Weight gradient (dj_dw)         : {dj_dw}")
```

Since y = 1 and the predicted probability ŷ = 0.9526 is slightly less than 1, the model needs to increase its prediction. The computed weight gradients are negative (dj_dw = [-0.0711, -0.0237]). According to the gradient descent update rule, w = w − α·dj_dw, subtracting a negative gradient increases the weights. Larger weights increase z = W·X + b, which in turn increases the sigmoid output. Therefore, the model should increase the weights so that future predictions move closer to the actual label y = 1.

```python
prob_list = np.array([0.1, 0.4, 0.6, 0.9, 0.45])
y = np.array([0, 0, 1, 1, 0])
thresold = 0.5

predictions = ( prob_list > thresold )

y_pred = predictions.astype(int)

# print(y_pred)

correct_predictions = np.sum( y_pred == y )

accuracy = correct_predictions / len(y)

false_pos = np.sum( ( y_pred == 1 ) & ( y == 0 ) )
false_neg = np.sum( ( y_pred == 0 ) & ( y == 1 ) )

print(f'Predicted classses : {y_pred}')
print(f'Accuracy          : {accuracy * 100:.1f}%')
print(f'False Positive    : {false_pos}')
print(f'False Negative    : {false_neg}')
```

# Problem 06

### a) Why is accuracy misleading here?

The dataset contains 95% negative samples and only 5% positive samples.
If the model predicts every sample as class 0 (negative), it will still achieve 95% accuracy because most samples belong to the negative class.

However, the model completely fails to identify any positive samples. Therefore, accuracy gives a false impression that the model is performing well.

---

### b) Which evaluation metric(s) would you use instead?

For imbalanced datasets, the following metrics are more informative:

- Precision: Measures how many predicted positive samples are actually positive.
- Recall (Sensitivity): Measures how many actual positive samples are correctly detected.
- F1-Score: Harmonic mean of Precision and Recall.
- ROC-AUC Score: Evaluates the model's ability to distinguish between classes.
- Precision-Recall Curve (PR-AUC): Particularly useful for highly imbalanced datasets.

Recall and F1-Score are usually the most important metrics when detecting the minority class.

---

### c) How can you adjust the model to detect more positive samples?

To improve detection of positive samples, we can:

1. Use class weights (e.g., class_weight='balanced') to penalize mistakes on the minority class more heavily.
2. Lower the classification threshold (e.g., use 0.3 instead of 0.5).
3. Oversample the minority class using techniques such as SMOTE.
4. Undersample the majority class.
5. Collect more positive-class data if possible.

These techniques help the model pay more attention to the minority (positive) class and improve recall.

## b) predict_probability(X, w, b)

Function predict_probability(X, w, b):

    m = number of training examples
    predictions = empty list

    For i = 0 to m-1:

        z = dot(X[i], w) + b
        p = sigmoid(z)

        store p in predictions

    return predictions

--------------------------------------------------

## c) compute_cost(X, y, w, b)

Function compute_cost(X, y, w, b):

    m = number of training examples
    cost = 0

    For i = 0 to m-1:

        z = dot(X[i], w) + b
        p = sigmoid(z)

        cost += -y[i] * log(p)
                - (1 - y[i]) * log(1 - p)

    cost = cost / m

    return cost

--------------------------------------------------

## d) compute_gradients(X, y, w, b)

Function compute_gradients(X, y, w, b):

    m = number of training examples
    n = number of features

    dj_dw = zeros(n)
    dj_db = 0

    For i = 0 to m-1:

        z = dot(X[i], w) + b
        p = sigmoid(z)

        error = p - y[i]

        For j = 0 to n-1:
            dj_dw[j] += error * X[i][j]

        dj_db += error

    dj_dw = dj_dw / m
    dj_db = dj_db / m

    return dj_dw, dj_db

--------------------------------------------------

## e) gradient_descent(X, y, w_init, b_init, alpha, iterations)

Function gradient_descent(X, y, w_init, b_init, alpha, iterations):

    w = w_init
    b = b_init

    For iter = 1 to iterations:

        dj_dw, dj_db = compute_gradients(X, y, w, b)

        For j = 0 to number_of_features - 1:
            w[j] = w[j] - alpha * dj_dw[j]

        b = b - alpha * dj_db

    return w, b

--------------------------------------------------

Summary of Formulas:

1. z = X·w + b

2. Sigmoid:
   p = 1 / (1 + e^(-z))

3. Cost:
   J = (1/m) Σ[-y log(p) - (1-y) log(1-p)]

4. Gradients:
   dj_dw = (1/m) Σ[(p - y)X]
   dj_db = (1/m) Σ(p - y)

5. Gradient Descent Update:
   w = w - α * dj_dw
   b = b - α * dj_db
